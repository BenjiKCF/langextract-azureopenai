```python
import langextract as lx
import textwrap
import dotenv
import os

dotenv.load_dotenv('.env.local')
```

```python
# 1. Define the prompt and extraction rules
prompt = textwrap.dedent("""\
    Extract characters, emotions, and relationships in order of appearance.
    Use exact text for extractions. Do not paraphrase or overlap entities.
    Provide meaningful attributes for each entity to add context.""")
```

```python
# 2. Provide a high-quality example to guide the model
examples = [
    lx.data.ExampleData(
        text="ROMEO. But soft! What light through yonder window breaks? It is the east, and Juliet is the sun.",
        extractions=[
            lx.data.Extraction(
                extraction_class="character",
                extraction_text="ROMEO",
                attributes={"emotional_state": "wonder"}
            ),
            lx.data.Extraction(
                extraction_class="emotion",
                extraction_text="But soft!",
                attributes={"feeling": "gentle awe"}
            ),
            lx.data.Extraction(
                extraction_class="relationship",
                extraction_text="Juliet is the sun",
                attributes={"type": "metaphor"}
            ),
        ]
    )
]
```

```python
# Sample text for extraction - a Romeo and Juliet excerpt
input_text = """ROMEO. But soft! What light through yonder window breaks? 
It is the east, and Juliet is the sun. 
Arise, fair sun, and kill the envious moon,
Who is already sick and pale with grief,
That thou her maid art far more fair than she.
Be not her maid, since she is envious;
Her vestal livery is but sick and green
And none but fools do wear it; cast it off.
JULIET appears above at a window.
It is my lady, O, it is my love!
O, that she knew she were!"""
```

```python
# # Load and register our Azure OpenAI provider
# import langextract as lx
# from langextract_azureopenai import AzureOpenAILanguageModel

# # Manually register the provider if not automatically loaded
# lx.providers.load_plugins_once()

# # Check if our provider is available
# print("Available provider patterns:")
# for entry in lx.providers.registry.list_entries():
#     print(f"  {entry}")
```

```python
# Run the extraction using Azure OpenAI GPT-5 provider
result = lx.extract(
    text_or_documents=input_text,
    prompt_description=prompt,
    examples=examples,
    model_id=f"azureopenai-{os.getenv('AZURE_OPENAI_DEPLOYMENT_NAME')}",  # Use our Azure OpenAI provider
    debug=False
)
```
